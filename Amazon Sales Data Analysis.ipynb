### Visualization questions

#### 1) Is there a relationship between product rating and number of reviews?

```python
import matplotlib.pyplot as plt

# Remove missing values

plot_data = df[['Rating', 'Number_of_reviews']].dropna()

# Create scatter plot

plt.figure(figsize=(10, 6))

plt.scatter(
    plot_data['Rating'],
    plot_data['Number_of_reviews'],
    alpha=0.4
)

plt.yscale('log')

plt.xlabel('Rating')
plt.ylabel('Number of Reviews (log scale)')
plt.title('Relationship Between Product Rating and Number of Reviews')

plt.show()
```

##### The scatter plot shows that there is no strong direct relationship between product rating and number of reviews. Products with very high review counts generally have ratings around 4.4–4.8, but products with lower review counts can also have high ratings. Therefore, a high number of customer reviews does not necessarily result in a higher rating. Review volume appears to vary much more widely than rating.

### 2) Do sponsored products have more customer reviews than organic products?

```python
import matplotlib.pyplot as plt

# Calculate average number of reviews for sponsored and organic products

sponsored_reviews = (
    df.groupby('is_sponsored')['Number_of_reviews']
      .mean()
      .sort_values(ascending=False)
)

# Create bar chart

plt.figure(figsize=(8, 5))

plt.bar(
    sponsored_reviews.index,
    sponsored_reviews.values
)

plt.xlabel('Product Type')
plt.ylabel('Average Number of Reviews')
plt.title('Average Reviews: Sponsored vs Organic Products')

plt.tight_layout()
plt.show()
```

##### Sponsored products have an average of approximately 3,706 reviews, compared with approximately 2,933 reviews for organic products. Therefore, sponsored products in this dataset have about 26% more average reviews than organic products. However, this does not prove that sponsorship causes more reviews, because other factors such as product popularity, brand, price, and product category may also influence review volume.

### 3) Is there a relationship between product price and customer rating?

```python
import matplotlib.pyplot as plt

# Select required columns and remove missing values

price_rating = df[
    ['Current/discounted_price', 'Rating']
].dropna()

# Create scatter plot

plt.figure(figsize=(10, 6))

plt.scatter(
    price_rating['Current/discounted_price'],
    price_rating['Rating'],
    alpha=0.4
)

plt.xlabel('Current / Discounted Price')
plt.ylabel('Rating')
plt.title('Relationship Between Product Price and Rating')

plt.tight_layout()
plt.show()
```

##### The scatter plot is expected to show no strong relationship between product price and rating. Products across a wide range of prices can have similar ratings, with ratings generally concentrated around 4.0–5.0. Therefore, a higher product price does not necessarily correspond to a higher customer rating.

### 4) Do products with higher discount percentages receive more reviews?

### 6) How are discount percentages distributed across products?

### 7) How does the average rating vary across different levels of monthly purchases?

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create demand categories

df['purchase_category'] = pd.cut(
    df['bought_in_last_month'],
    bins=[-1, 99, 999, 9999, float('inf')],
    labels=['Low', 'Medium', 'High', 'Very High']
)

# Calculate average rating for each purchase category

purchase_rating = (
    df.groupby('purchase_category', observed=False)['Rating']
      .mean()
)

# Create bar chart

plt.figure(figsize=(9, 5))

plt.bar(
    purchase_rating.index,
    purchase_rating.values
)

plt.xlabel('Monthly Purchase Category')
plt.ylabel('Average Rating')
plt.title('Average Rating Across Monthly Purchase Categories')

plt.ylim(0, 5)

plt.tight_layout()
plt.show()
```

##### The chart compares customer ratings across different levels of monthly purchase volume. The average ratings across the purchase categories are relatively close, indicating that products with higher monthly demand do not necessarily have substantially higher ratings. Therefore, customer demand and customer satisfaction appear to be related only weakly, and high sales volume cannot be explained by rating alone.
